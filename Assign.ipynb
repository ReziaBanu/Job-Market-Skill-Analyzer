```python
job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "ML_Engineer"],
    "Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
    "Skills": [
        "Excel SQL Python",
        "Python ML Statistics",
        "Excel SQL PowerBI",
        "Python SQL Spark",
        "Python ML DeepLearning"
    ]
}
print(job_data)

## TASK-1
for i in range(len(job_data["Job_ID"])):
    print(f" ID : {job_data['Job_ID'][i]}"
          f"   Desig  : {job_data['Role'][i]}"
          f"     Company: {job_data['Company'][i]}"
          f"     Skills: {job_data['Skills'][i]}")

## TASK-2
New_requirement = int(input("How many job postings you require: "))
for i in range(New_requirement):
    role = input("Enter role : ")
    company = input("Enter company : ")
    skills = input("Enter skills : ")

    upcoming_id = max(job_data["Job_ID"]) + 1

    job_data["Job_ID"].append(upcoming_id)
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

print(job_data)

## TASK-3
def clean_skills(skills):
    skills = skills.lower()
    skills = skills.replace(",", " ")
    skills = skills.split()
    return skills

job_data["Skills"] = [clean_skills(x) for x in job_data["Skills"]]

for x in job_data["Skills"]:
    print(f" Cleaned_Skill is : {x} ")

## TASK-4
Count = {}
for j in job_data["Skills"]:
    # j is already a list of skills, so we iterate through it directly without .split()
    for J in j:
        if J in Count:
            Count[J] += 1
        else:
            Count[J] = 1

for J in Count:
    print(f" Skills are : {J} , {Count[J]} ")

## TASK-5
print(f"  Total no of job_postings : {len(job_data['Role'])}")

unique_skills = set(Count.keys())
print(f" Unique skills are : {unique_skills}")

tot_jobs = len(job_data["Job_ID"])
python_jobs = []
for i in range(tot_jobs):
    # job_data["Skills"][i] is already a list, so we do not use .split()
    if "python" in job_data["Skills"][i]:
        python_jobs.append(job_data["Role"][i])

print(f" Jobs that mentioning Python are: {python_jobs}")

##  TASK-6

def find_jobs_by_skill(skill_query):
    skill_query = skill_query.lower()
    matching_jobs = []

    for i in range(len(job_data["Skills"])):
        job_skills = job_data["Skills"][i] # Already a list

        if skill_query in job_skills:
            matching_jobs.append((job_data["Job_ID"][i],
                                 job_data["Role"][i],
                                 job_data["Company"][i]))
    return matching_jobs

query = input("Enter skill query: ")
results = find_jobs_by_skill(query)
if results:
    print(f"Matching jobs: {results}")
else:
    print("No matching jobs found.")

## TASK-7

# Define a lambda function to check if a skill is common
common_skills = lambda skill: Count[skill] >= 2

roles_with_common_skills = [
    job_data["Role"][i]
    for i in range(len(job_data["Role"]))
    if any(common_skills(skill) for skill in job_data["Skills"][i])
]

print(f" Roles with at least one common skill: {roles_with_common_skills}")

##  TASK - 8

unique_skills = set(Count.keys())
print(f" Unique skills are : {sorted(unique_skills)}")

##  TASK - 9


import csv

try:
  with open("job_skills.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Job_ID", "Role", "Company", "Skills"])
    for i in range(len(job_data["Job_ID"])):
      writer.writerow([ job_data["Job_ID"][i],
                        job_data["Role"][i],
                        job_data["Company"][i],
                        job_data["Skills"][i]])

except OSError as e:
    print("Error saving CSV file:", e)

else:
    print("job_skills.csv saved successfully.")

finally:
    print("CSV file operation completed.")

```
